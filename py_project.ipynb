```python
%pip install pyfiglet
from datetime import datetime,timedelta
import pyfiglet
import math
import random
tasks=[]
```

show summary

```python
def show_summary():
  if tasks==[]:
    print("no tasks yet")
    return

  count_of_complete_tasks=0
  count_of_pending_tasks=0
  total_Estimated_minutes=0
  uniq_tasks=set()
  for i in range(len(tasks)):
    uniq_tasks.add(tasks[i]["Course"])
    if tasks[i]["comlation"]==False:
      count_of_pending_tasks+=1
      total_Estimated_minutes+=tasks[i]["Estimated minutes"]
    else:
      count_of_complete_tasks+=1
  progress=100*(count_of_complete_tasks/len(tasks))
  print(f"total number of tasks is {len(tasks)}")
  print(f"total number of completed tasks is {count_of_complete_tasks}")
  print(f"total number of pending tasks is {count_of_pending_tasks}")
  print(f"your progres percentage is {progress:.1f}%")
  print(f"you need {total_Estimated_minutes} to finish")
  print(uniq_tasks)
```

Add a task

```python
def Add_a_task():
  show_summary()
  task={}
  while True:
    try:
        task["Course"]=(input("please enter the course name:  ")).strip()
        if task["Course"]=="":
          continue
        else:
          break
    except ValueError:
        print("Please enter a string:  ")
  while True:
    try:
      task["Task"]=str(input("please enter the Task:  ")).strip()
      if task["Task"]=="":
        continue
      else:
          break
    except ValueError:
        print("Please enter a string")
  while True:
    try:
      task["Estimated minutes"]=int(input("please enter the Estimated minutes and it must be greater than zero|:  "))
      if task["Estimated minutes"]>0:
        break
      else:
        continue
    except ValueError:
        print("Please enter a number:  ")
  while True:
    try:
      task["Due in days"]=int(input("please enter the Due in days and it must be zero or positive:  "))
      if task["Due in days"]>=0:
        break
      else:
        continue
    except ValueError:
        print("Please enter a number:  ")
  task["Course"]=task["Course"].strip()
  task["Task"]=task["Task"].strip()
  target_time=datetime.now()+timedelta(days=task["Due in days"])
  task["appoitment"]=target_time.strftime("%d/%m/%Y")
  task["comlation"]=False
  return(task)
```

view tasks


```python
def view_tasks():
  if tasks==[]:
    print("no tasks yet")
    return
  for i in range(len(tasks)):
      print(f"""{i+1}. {tasks[i]["Course"]} -- {tasks[i]["Task"]}
Estimated minutes: {tasks[i]["Estimated minutes"]}
Due date: {tasks[i]["appoitment"]}""")
      if tasks[i]["comlation"]==False:
        print("Status: Pending")
      else:
        print("Status: Complete")

```

Complete task

```python
def Complete_task():
  if tasks==[]:
    print("no tasks yet")
    return
  for i in range(len(tasks)):

    print(f'{i+1}. {tasks[i]["Course"]}  {tasks[i]["Task"]}')
    if tasks[i]["comlation"]==False:
      print("Status: Pending")
    else:
      print("Status: Complete")
  while True:
      try:
        completed_task=int(input("enter the task you completed"))
        if completed_task<=len(tasks) and completed_task>0:
          if tasks[completed_task-1]["comlation"]==True:
            print("alredy compleated")
            break
          else:
            tasks[completed_task-1]["comlation"]=True
            break
        else:
          print("chose the right task number")
      except ValueError:
          print("enter a valid number")
```

suggust task

```python
def suggust_task():
  uncompleated_task=[]
  session_lengths = (25, 45, 60)
  for i in range(len(tasks)):
    if tasks[i]["comlation"]==False:
      uncompleated_task.append(tasks[i])
    else: continue
  if uncompleated_task==[]:
    print(pyfiglet.figlet_format("Will Done you completed all your tasks"))
  else:
    suggusted_task=random.choice(uncompleated_task)
    print(f"I suggest you go with {suggusted_task["Course"]} {suggusted_task["Task"]}")
    while True:
      try:
        session_length=int(input(f"chose the session_length {session_lengths}"))
        if session_length in session_lengths:
           session_needed=math.ceil(suggusted_task["Estimated minutes"]/session_length)
           print(f"yoy need {session_needed} sesions")
           break
        else:
          print("chose a right session_length")
      except ValueError:
          print("enter a valid number")
```
